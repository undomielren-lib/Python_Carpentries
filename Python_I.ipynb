## Variables and Assignment

Python does not store values unless we assign variables.

- Variable names can only contain letters, digits, underscores
- Cannot start with a digit
- Case sensitive
- Should be meaningful
- `=` assigns the value on the right to the name on the left

## Print

Python has built-in functions. The `print` function displays values.

- Call the function (tell Python to run it) by using its name
- Provide values to the function (the things to print) in parentheses: `function()`
- Add text to print with strings using single or double quotes
- Values passed to a function are called **arguments**

## Order Matters

- Order of execution takes precedence over order of appearance
- Python remembers **all** previously run code (including variables)
- If things get out of order or stuck, use **Kernel → Restart & Run All** to clear the interpreter

## Indexing

- Characters in a string are ordered (a list of characters)
- Each position in the string is assigned a number, called an **index** or **subscript**
- Python uses zero-based indexing — the first element of a list, tuple, or string is always at index `0`
- Use the position's index in square brackets to get the character at that position

## Slicing

- **Substring**: part of a string
- **Element**: item in a list
- **Slice**: part of a string or any list-like thing — `[start:stop]`
    - `start` is the integer index of the first element we want
    - `stop` is the integer index of the element just after the last element we want
    - The difference between `start` and `stop` is the slice's length

## Challenge

```python
species_name = 'Acacia buxifolia'
```

- `species_name[2:8]`
- `species_name[11:]` (without a value after the colon)
- `species_name[:4]` (without a value before the colon)
- `species_name[:]` (just a colon)
- `species_name[11:-3]`
- `species_name[-5:-3]`

What happens when you choose a stop value which is out of range? (i.e., try `species_name[0:20]` or `species_name[:103]`)

## Slicing Solution

- `species_name[2:8]` returns the substring `'acia b'`
- `species_name[11:]` returns the substring `'folia'`, from position 11 until the end
- `species_name[:4]` returns the substring `'Acac'`, from the start up to but not including position 4
- `species_name[:]` returns the entire string `'Acacia buxifolia'`
- `species_name[11:-3]` returns the substring `'fo'`, from the 11th position to the third-last position
- `species_name[-5:-3]` also returns the substring `'fo'`, from the fifth-last position to the third-last position
- If a part of the slice is out of range, the operation does not fail: `species_name[0:20]` gives the same result as `species_name[0:]`, and `species_name[:103]` gives the same result as `species_name[:]`

## Data Types & Type Conversion

Every value has a specific type:

- **Integer** (`int`): positive or negative whole numbers, like `3` or `-512`
- **Floating point number** (`float`): real numbers, like `3.14159` or `-2.5`
- **Character string** (`str`): text
    - Single or double quotes, as long as they match
    - Quotes aren't displayed when printed
- Use the `type` function to find the type of a value!

## More on Types

- Types control what operations (or methods) can be performed on a given value
- You can use `+` and `*` operators on strings
    - `+` concatenates them
    - Multiplying a character string by integer `N` creates a new string with that character repeated `N` times
- Strings have a length (but numbers don't)

## Converting Numbers to Strings (or Vice Versa)

- Cannot add numbers and strings — it's ambiguous
- Some types can be converted to others by using the type name as a function
- You can mix integers and floats freely in operations
    - Int and float can be mixed in arithmetic
    - Float always wins

## Built-in Functions

- Use `#` to add comments
- A function may take zero or more arguments
    - `len` takes exactly one
    - `int`, `str`, `float` create a new value from an existing one
    - `print` takes zero or more
- Commas separate arguments
- Common ones: `max`, `min`, and `round`

## Methods

- Methods are functions attached to objects
- Have parentheses like functions, but come after the variable
- Can be chained together, operating left to right

## Help

Every built-in function has online documentation, e.g. `help(round)`.

- **Option 1:** place the cursor near where the function is invoked in a cell
    - Hold Shift, press Tab
    - Repeat to expand the information returned
- **Option 2:** type the function name in a cell with a question mark after it and run the cell

## Errors

- **Syntax error**: the program won't even try to run if it can't be parsed
    - Fix by reading the source code
- **Runtime error**: something goes wrong while a program is executing
    - Fix by tracing execution

## Libraries

- A **library** (module): a collection of files that contain functions for use by other programs
    - Also data values (e.g. numerical constants)
- Use `import` to load a library
    - Access with `module_name.thing_name` — Python uses a period to mean "part of"
- Use `from ... import ...` to load specific items and refer to them directly, without the library name as a prefix
- Create an alias to shorten the name: `mpl` (matplotlib), `m` (math), `pd` (pandas), `np` (numpy), `plt` (pyplot)

## Reading Tabular Data

- **Pandas** is a Python library for statistics, particularly tabular data
    - Borrows from R's dataframes
- Load pandas: `import pandas as pd`
- Read a CSV (Comma Separated Values) data file with `pd.read_csv`
    - Argument is the name of the file
    - Returns a **dataframe** that you can assign to a variable
